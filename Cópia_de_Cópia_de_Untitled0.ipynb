```python
# Sistema Bancario

# Funções Principais do Sistema

# Função de Consulta do Crédito Por Score
def consultar_credito():
  try:
    score = int(input("Digite o score de crédito do cliente: "))

    if score > 0 and score < 300:
      print("Classificação: Crédito Negado.")

    elif score < 600:
      print("Classificação: Crédito Restrito.")

    elif score < 800:
      print("Classificação: Crédito Aprovado.")

    elif score <= 1000:
      print("Classificação: Crédito Premium.")

    else:
      print("Score inválido. Tente novamente.")

  except ValueError:
    print("Informe apenas números. Tente novamente.")

# Função de Consultar Conta por Renda
def consultar_conta():
  try:
    renda = int(input("Informe a renda mensal do cliente: "))

    if renda > 0 and renda < 2001:
      print("Categoria: Conta Básica.")

    elif renda < 5001:
      print("Categoria: Conta Especial.")

    elif renda < 15001:
      print("Categoria: Conta Premium")

    elif renda >= 15001:
      print("Categoria: Conta Private.")

    else:
      print("Valor negativo não é aceito. Tente novamente.")

  except ValueError:
    print("Use apenas números. Tente novamente.")

# Função Para Consultar Atendimento Por Idade e Canal de Preferência
def consultar_atendimento():
  try:
    idade = int(input("Informe a idade do cliente: "))
    canal = input("Informe o canal de atendimento (agência/ aplicativo/ telefone) [Obs - Digite da mesma forma apresentada]: ")

    if idade < 0:
      print("Idade inválida. Tente novamente.")

    elif idade >= 65 and canal == "aplicativo":
      print("Devido a idade do cliente, recomendamos os canais de telefone ou agência para um melhor atendimento.")

    elif canal == "agência" or canal == "telefone" or canal == "aplicativo":
      print(f'O canal de atendimento escolhido foi: {canal}.')

    else:
      print("Canal inválido. Escolha entre uma das opções (agência/ aplicativo/ telefone).")

  except ValueError:
    print("Tente novamente.")

# Laço de Repetição Para o Funcionamento Contínuo
while True:

# Menu Inicial
  print("===== Central de Atendimento Bancário =====")
  print(" 1 - Consultar classificação de crédito (por score)")
  print(" 2 - Consultar categoria de conta (por faixa de renda)")
  print(" 3 - Consultar canal de atendimento recomendado (por idade e canal de preferência)")
  print(" 0 - Sair")

  opcao = input("Digite uma das opções (0-3): ")

  match opcao:

    case "1":
      consultar_credito()
      break

    case "2":
      consultar_conta()
      break

    case "3":
      consultar_atendimento()

    case "0":
      print("Atendimento Encerrado!")
      break

    case _:
      print("Opção inválida. Tente novamente.")
      continue


```
